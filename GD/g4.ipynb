```python
import numpy as np
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=9)
```

```python
M1 = LinearRegression()
M1.fit(X_train,y_train)
print(M1.coef_)
print(M1.intercept_)
```

```python
y_pred = M1.predict(X_test)
r2_score(y_test,y_pred)
```

```python
class BGDRegressor:

    def __init__(self,learning_rate=0.01,epcohs=100):
        self.intercept_ = None
        self.coef_ = None
        self.LR=learning_rate
        self.EP=epcohs

    def fit(self,X_train,y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.EP):
            y_pred = self.intercept_ + np.dot(X_train,self.coef_)

            intercept_der = -2 * np.mean(y_train - y_pred)
            self.intercept_ = self.intercept_ - self.LR*intercept_der

            coef_der = -2 * np.dot((y_train - y_pred),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - self.LR *coef_der
        print(self.coef_)    
        print(self.intercept_)    

    

    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python

m2 = BGDRegressor(learning_rate=0.01,epcohs=50)
import time

start = time.time()
m2.fit(X_train,y_train)
print('Taken time :',time.time()-start)
```

```python
y_pred = m2.predict(X_test)
```

```python
r2_score(y_test,y_pred)        
```
