## Fitting a Linear-Model

```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# Inputs: Hour of studies and Marks 

X = np.array([[1],[2],[3],[4],[5],[6]])
y = np.array([55,60,65,70,75,80])

model = LinearRegression()
model.fit(X,y)
```

```python
print("Weight (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
model.predict([[7]])
```

```python
prediction = model.predict([[4]])
prediction[0]
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split

X_tr , X_ts , y_tr , y_ts = train_test_split(X,y,test_size=0.2,random_state=8)
```

```python
from sklearn.linear_model import LinearRegression

model3 = LinearRegression()
model3.fit(X_tr,y_tr)
```

```python
X_ts.head()
```

```python
y_ts.head()
```

```python
model3.predict(X_ts.iloc[[0]])

# Coz we needed data frame for X
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_tr,model3.predict(X_tr),color='red')
plt.xlabel('CGPA')
plt.ylabel('PACKAGE')
```
