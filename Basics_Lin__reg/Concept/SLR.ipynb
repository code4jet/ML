## ML class from Scratch

```python
class myLR :
    def __init__(self):
        self.m = None
        self.b = None

    def fit(self,X_train,Y_train):

        num = 0
        den = 0

        X_train_mean = X_train.mean()
        Y_train_mean = Y_train.mean()

        for i in range(X_train.shape[0]):

            num = num + ((X_train[i] - X_train_mean)*(Y_train[i] - Y_train_mean))
            den = den + ((X_train[i] - X_train_mean)*(X_train[i] - X_train_mean))

        self.m = num/den
        self.b = Y_train.mean() - (self.m*X_train.mean())
        print(self.m)
        print(self.b)
       
    def predict(self,X_test):
        
        return self.m * X_test + self.b        
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('placement.csv')
df.head()
```

```python
X = df.iloc[:,0].values
y = df.iloc[:,-1].values
y
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test ,y_train ,y_test = train_test_split(X,y,test_size=0.2,random_state=4)

```

```python
model  = myLR()
model.fit(X_train,y_train)
```

```python
import numpy as np

clean = np.array([10, 11, 12, 13, 14])
with_outlier = np.append(clean, 100)   # one extreme value added

# The mean is the best constant for MSE; the median is the best constant for MAE.
# We compare how much each one moves when the outlier appears.
print("Without outlier -> mean:", np.mean(clean), "| median:", np.median(clean))
print("With outlier    -> mean:", round(np.mean(with_outlier), 2), "| median:", np.median(with_outlier))
```

## Regression Metrics

```python
from sklearn.metrics import mean_absolute_error 
from sklearn.metrics import mean_squared_error 
from sklearn.metrics import root_mean_squared_error 
from sklearn.metrics import r2_score 


y_pred = model.predict(X_test.reshape(-1,1))
y_pred1 = model1.predict(X_test.reshape(-1,1))
print("MAE : ", mean_absolute_error(y_test,y_pred))
print("MAE : ", mean_absolute_error(y_test,y_pred1))
```

```python
print("MSE : ", mean_squared_error(y_test,y_pred))
print("MSE : ", mean_squared_error(y_test,y_pred1))
```

```python
print("MSE : ", root_mean_squared_error(y_test,y_pred))
print("MSE : ", root_mean_squared_error(y_test,y_pred1))
```

```python
print("MSE : ", r2_score(y_test,y_pred))
print("MSE : ", r2_score(y_test,y_pred1))
```
